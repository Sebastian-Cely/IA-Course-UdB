## <span style="color:#F97066;">1.</span> Vectores: operaciones básicas y norma

Un **vector** es una lista ordenada de números que representa un punto o una dirección en un espacio de $n$ dimensiones. En Machine Learning, cada ejemplo de un conjunto de datos suele representarse como un vector de características (*feature vector*): por ejemplo, una casa descrita por su área, número de habitaciones y antigüedad es el vector $[120, 3, 8]$.

A continuación, se definen dos vectores y se calculan algunas operaciones básicas, incluyendo su **norma** (magnitud o longitud del vector), que corresponde a la distancia euclidiana entre el vector y el origen. La norma es la base de métricas de distancia y similitud ampliamente usadas en Machine Learning (por ejemplo, en el algoritmo K-Vecinos Más Cercanos o en términos de regularización).

```python
import numpy as np

# Definir dos vectores
v1 = np.array([3, 4])
v2 = np.array([1, 2])

# Suma de vectores
print(f"Suma de vectores (v1 + v2): {v1 + v2}")

# Multiplicación por un escalar
print(f"v1 multiplicado por 2: {v1 * 2}")

# Norma (magnitud) de un vector: distancia euclidiana al origen
norma_v1 = np.linalg.norm(v1)
print(f"Norma de v1: {norma_v1}")
```

<a id="suma-resta"></a>

## <span style="color:#F97066;">2.</span> Suma y resta de matrices

Una **matriz** es una tabla de números organizada en filas y columnas; en Machine Learning, un conjunto de datos completo suele representarse como una matriz, donde cada fila es una observación y cada columna una variable. La suma y la resta de matrices son operaciones **elemento a elemento** (ambas matrices deben tener las mismas dimensiones) y aparecen constantemente al actualizar los parámetros de un modelo: por ejemplo, la regla de actualización del descenso de gradiente consiste, precisamente, en restarle a la matriz de pesos actual una fracción del gradiente.

```python
import numpy as np

# Definir matrices A y B
A = np.array([[2, 4], [1, 3]])
B = np.array([[5, 1], [2, 6]])

# Suma de matrices
suma = A + B
print(f"Suma de matrices (A + B):\n{suma}")

# Resta de matrices
resta = A - B
print(f"\nResta de matrices (A - B):\n{resta}")
```

<a id="producto-punto"></a>

## <span style="color:#F97066;">3.</span> Producto punto y producto matricial

El **producto punto** (o producto escalar) entre dos vectores multiplica sus componentes correspondientes y suma los resultados, produciendo un único número (escalar). Es la operación central detrás de la similitud coseno y de la salida de una neurona: cada neurona calcula, esencialmente, el producto punto entre sus entradas y sus pesos.

Cuando se aplica `np.dot()` sobre dos **matrices** (en lugar de dos vectores), la operación resultante es el **producto matricial**, que combina filas de la primera matriz con columnas de la segunda. A continuación se muestran ambos casos.

```python
# Producto punto entre dos vectores (resultado: un escalar)
producto_punto_vectores = np.dot(v1, v2)
print(f"Producto punto (v1 · v2): {producto_punto_vectores}")
```

```python
# Producto matricial
producto_matrices = np.dot(A, B)
print(f"\nProducto matricial (A·B):\n {producto_matrices}") 
```

<a id="matriz-vector"></a>

## <span style="color:#F97066;">4.</span> Producto matriz-vector

El caso particular del producto matricial más usado en Machine Learning es el **producto de una matriz por un vector**: $y = Ax$. Esta operación representa una **transformación lineal** aplicada al vector $x$, y es exactamente lo que calcula una capa de una red neuronal (antes de la función de activación) o la predicción de un modelo de regresión lineal ($\hat{y} = Xw$).

<span style="background:#0EA5E9;color:#ffffff;padding:4px 14px;border-radius:14px;font-size:0.75em;">🧪 EJERCICIO</span>

Ejecute la siguiente celda y observe cómo la matriz `A` transforma al vector `x` en un nuevo vector.

```python
# Vector de entrada
x = np.array([1, 2])

# Producto matriz-vector: y = A x
y = A.dot(x)
print(f"A · x = {y}")
```

<a id="transpuesta-inversa"></a>

## <span style="color:#F97066;">5.</span> Transposición e inversa de una matriz

La **transpuesta** de una matriz intercambia filas por columnas; se utiliza, por ejemplo, para adaptar las dimensiones de dos matrices antes de multiplicarlas. La **inversa** de una matriz $A$ (denotada $A^{-1}$) es aquella que, multiplicada por $A$, produce la matriz identidad; solo existe si $A$ es cuadrada y su determinante es distinto de cero.

Ambas operaciones son la base de la **ecuación normal**, la fórmula cerrada que resuelve una regresión lineal sin necesidad de descenso de gradiente: $w = (X^TX)^{-1}X^Ty$, donde $X^T$ es la transpuesta de la matriz de datos y $(X^TX)^{-1}$ es la inversa del producto matricial resultante.

```python
# Transpuesta de A
transpuesta_A = A.T
print(f"\nTranspuesta de A:\n{transpuesta_A}")

# Inversa de A (si es invertible)
if np.linalg.det(A) != 0:  # Verificamos que sea invertible
    inversa_A = np.linalg.inv(A)
    print(f"\nInversa de A:\n{inversa_A}")
else:
    print("\nLa matriz A no es invertible")
```
